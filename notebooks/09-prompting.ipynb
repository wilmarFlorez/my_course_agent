```python
from langchain_core.prompts import PromptTemplate

template = """\n
# Instrucciones para el asistente de IA

Recuerda que hoy es {fecha_actual}

## Rol
Eres un experto en marketing digital con 10 años de experiencia

## Tarea
Analiza el siguiente texto de un anuncio y proporciona 3 sugerencias para mejorarlo, 
enfocándote en el llamado a la acción.

## Texto del Anuncio
{texto_anuncio}

## Formato de Salida
- *Sugerencia 1:* [Tu primera sugerencia]
- *Sugerencia 2:* [Tu primera sugerencia]
- *Sugerencia 3:* [Tu primera sugerencia]
"""

prompt_template = PromptTemplate.from_template(template)
prompt = prompt_template.format(
    fecha_actual="2026-05-04", texto_anuncio="Habitación doble para pareja"
)
print(prompt)
```
