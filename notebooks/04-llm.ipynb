```python
from dotenv import load_dotenv

load_dotenv()
```

```python
from langchain_openai import ChatOpenAI
```

```python
gpt_5_4_mini = ChatOpenAI(model="gpt-5.4-mini", temperature=0)
response = gpt_5_4_mini.invoke("Hello, how are you?")
response.pretty_print()
```

```python
gpt_5_4_nano = ChatOpenAI(model="gpt-5.4-nano", temperature=0)
response = gpt_5_4_nano.invoke("Hello world")
response.pretty_print()
```

```python
from langchain_core.messages import AIMessage, HumanMessage, SystemMessage

msg_1 = SystemMessage(content="You are a helpful assistant that talks in spanish")
msg_2 = HumanMessage(content="Me llamo Wilmar")
msg_3 = AIMessage(content="Hola Wilmar, cómo estás?")
msg_4 = HumanMessage(content="Cómo me llamo?")
history = [msg_1, msg_2, msg_3, msg_4]
response = gpt_5_4_nano.invoke(history)
response.pretty_print()
```

```python
from langchain_anthropic import ChatAnthropic

claude_llm = ChatAnthropic(model="claude-haiku-4-5", temperature=0)
response = claude_llm.invoke(history)
response.pretty_print()
```

```python
from langchain.chat_models import init_chat_model

llm = init_chat_model("claude-haiku-4-5", temperature=0)
response = llm.invoke(history)
response.pretty_print()
```
