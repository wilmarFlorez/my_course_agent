```python
from langchain.chat_models import init_chat_model

llm = init_chat_model("claude-haiku-4-5", temperature=0)
response = llm.invoke("Hola mi nombr es Wilmar")
```

```python
from pydantic import BaseModel, Field


class ContactInfo(BaseModel):
    """Contact info for a guest"""

    name: str = Field(description="The name of the guest")
    email: str = Field(description="The email address of the guest")
    phone: str = Field(description="The phone number of the guest")
    tone: int = Field(description="The tone of the person", ge=0, le=100)
    setiment: str = Field(description="The sentiment converssation of the guest")


llm_with_structured_output = llm.with_structured_output(schema=ContactInfo)

messages = [
    (
        "system",
        """you are a helpful assistant that can extract cotact 
     information from a given conversation.""",
    ),
    ("user", "Hola soy wilmar mi número de teléfono es +57 318 433 44428"),
]

response = llm_with_structured_output.invoke(messages)
response
```
